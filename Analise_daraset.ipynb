# Analise Estatística

```python
# Qual a Proporção de pessoas que tiveram Problema?

print(data['has_heart_disease'].map({0:"Não", 1:"Sim"}).value_counts(normalize= True))


data['has_heart_disease'].map({0:"Não", 1:"Sim"}).value_counts(normalize= True).plot(kind= "bar")
```

```python
# faixa etária com maior risco?

faixas_etarias = np.arange(18, 91, 5)

fig, axes = plt.subplots(2, 1, figsize= (10, 10))

ax = axes.flat[0]

data["faixas_etaria"] = pd.cut(data["age"], bins= faixas_etarias)

data_faixas = data.groupby("faixas_etaria", observed= True)["has_heart_disease"].mean()*100

ax.bar(data_faixas.index.astype(str) , data_faixas.values, color = '#C44E52')

# Sexo com maior risco

SEX_COLORS = {"Male": '#4C72B0', "Female": '#DD8452'}
ax = axes.flat[1]

data_sexo = data.groupby("sex")["has_heart_disease"].mean()*100

cores = [SEX_COLORS[i] for i in data_sexo.index]

data_sexo.plot(kind= "bar", ax= ax, color= cores)

plt.tight_layout()
```

```python
# Diferença por Caracteristica
clinical_fields = ['resting_bp_systolic','cholesterol_total','ldl','hdl','bmi','hba1c']
DISEASE_COLORS = {0: 'green', 1: 'red'}

fig, axes = plt.subplots(2, 3, figsize = (17, 5))


for i in range(len(clinical_fields)):
    ax = axes.flat[i]
    field = clinical_fields[i]
    title = clinical_fields[i]

    for status, c in DISEASE_COLORS.items():
        ax.hist(data[data.has_heart_disease == status][field], bins= 25, alpha = 0.6, color= c , density= True)
        ax.set_title(title)
plt.suptitle('Clinical Measurement Distributions — Green = No Disease, Red = Has Disease')
plt.tight_layout()
plt.show()
```

```python
# A razão HDL/LDl é melhor do ela isolada?

data["ldl_hdl_ratio"] = (data["ldl"]/ data["hdl"]).round(2)

q = pd.qcut(data["ldl_hdl_ratio"], 8)

ax = (data.groupby(q, observed= True)["has_heart_disease"].mean()*100).plot(marker= "o")

ax.axhline(0.30 * 100, color='black', ls='--', label='overall prevalence')
ax.set_xlabel("LDL / HDL ratio (octile)")
plt.setp(ax.get_xticklabels(), rotation=35, ha='right')
ax.legend()
plt.tight_layout()
plt.show()
```

```python
#correlação em relação as variaveis

variaveis = data.select_dtypes(np.number).drop(columns="patient_id")
correlacao = variaveis.corr(method= "pearson")["has_heart_disease"].drop("has_heart_disease").sort_values()
ax = correlacao.plot(kind= "barh", color = np.where(correlacao > 0, '#C44E52', '#4C72B0'))
plt.tight_layout()
plt.show()
```

```python
# Analisando variaveis categoricas

variaveis_categoricas = data.select_dtypes(exclude= np.number)

print("variaveis Categoricas:")
print("")
print(variaveis_categoricas.info())
```

```python
fig, axes = plt.subplots(2, 3, figsize = (15, 10))

# analise do varialvel "chest_pain_type"
ax = axes.flat[0]

analise_pain = data.groupby("chest_pain_type")["has_heart_disease"].mean()*100

analise_pain.plot(kind= "barh", ax= ax)
ax.bar_label(ax.containers[0], fmt='%.1f')
ax.set_title("chest_pain_type")
ax.set_xlabel("% with heart disease")

# analise do varialvel "exercise_induced_angina"
ax = axes.flat[1]

analise_pain = data.groupby("exercise_induced_angina")["has_heart_disease"].mean()*100

analise_pain.plot(kind= "bar", ax= ax)
ax.bar_label(ax.containers[0], fmt='%.1f')
ax.set_title("exercise_induced_angina")
ax.set_ylabel("% with heart disease")

# analise do varialvel "family_history"
ax = axes.flat[2]

analise_pain = data.groupby("family_history")["has_heart_disease"].mean()*100

analise_pain.plot(kind= "bar", ax= ax)
ax.bar_label(ax.containers[0], fmt='%.1f')
ax.set_title("family_history")
ax.set_ylabel("% with heart disease")

# analise do varialvel "smoker_status"
ax = axes.flat[3]

analise_pain = data.groupby("smoker_status")["has_heart_disease"].mean()*100

analise_pain.plot(kind= "bar", ax= ax)
ax.bar_label(ax.containers[0], fmt='%.1f')
ax.set_title("smoker_status")
ax.set_ylabel("% with heart disease")
# analise do varialvel "wearable_owner"
ax = axes.flat[4]

analise_pain = data.groupby("wearable_owner")["has_heart_disease"].mean()*100

analise_pain.plot(kind= "bar", ax= ax)
ax.bar_label(ax.containers[0], fmt='%.1f')
ax.set_title("wearable_owner")
ax.set_ylabel("% with heart disease")

plt.tight_layout()
```
